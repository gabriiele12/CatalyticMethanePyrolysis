```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import scienceplots

from matplotlib.ticker import LogLocator, FuncFormatter
from pathlib import Path


plt.style.use(["science", "ieee"])

tick_font = 13.5
xlabel_font = 13.0
axis_label_font = 14.8
value_font = 12.5

mpl.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Times"],
    "font.size": tick_font,
    "axes.labelsize": axis_label_font,
    "xtick.labelsize": xlabel_font,
    "ytick.labelsize": tick_font,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "text.color": "black",
    "axes.labelcolor": "black",
    "axes.edgecolor": "black",
    "xtick.color": "black",
    "ytick.color": "black",
    "axes.grid": False,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})


catalyst_labels = [
    "Ni\n600 $^\\circ$C",
    "Fe\n725 $^\\circ$C",
    "AC\n825 $^\\circ$C"
]

catalyst_amounts = np.array([
    950,
    10000,
    210000
])


catalyst_colors = [
    "#E69F00",  # Nickel
    "#7B4AB5",  # Iron
    "#D62728"   # Activated carbon
]


#Log scale
y_min = 500
y_max = 350000


fig, ax = plt.subplots(
    figsize=(5.6, 4.4)
)

fig.patch.set_facecolor("white")
ax.set_facecolor("white")

x_positions = np.arange(
    len(catalyst_labels)
)


#Bars
bars = ax.bar(
    x_positions,
    catalyst_amounts - y_min,
    bottom=y_min,
    width=0.60,
    color=catalyst_colors,
    edgecolor="black",
    linewidth=1.0,
    zorder=2
)

ax.set_yscale("log")

ax.set_ylim(
    y_min,
    y_max
)

ax.yaxis.set_major_locator(
    LogLocator(
        base=10,
        numticks=10
    )
)

ax.yaxis.set_minor_locator(
    LogLocator(
        base=10,
        subs=np.arange(2, 10) * 0.1,
        numticks=100
    )
)

ax.yaxis.set_major_formatter(
    FuncFormatter(
        lambda value, position: f"{value:,.0f}"
    )
)

ax.yaxis.set_minor_formatter(
    FuncFormatter(
        lambda value, position: ""
    )
)


for bar, value in zip(
    bars,
    catalyst_amounts
):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value * 1.025,
        f"{value:,.0f}",
        ha="center",
        va="bottom",
        fontsize=value_font,
        color="black"
    )

ax.set_ylabel(
    r"Catalyst amount / [kg$_{\mathrm{cat}}$]",
    fontsize=axis_label_font,
    labelpad=10
)

ax.set_xticks(
    x_positions
)

ax.set_xticklabels(
    catalyst_labels,
    fontsize=xlabel_font
)


ax.set_xlim(
    -0.65,
    len(catalyst_labels) - 0.35
)

ax.tick_params(
    axis="x",
    which="major",
    direction="in",
    top=True,
    bottom=True,
    length=4.9,
    width=1.1,
    labelsize=xlabel_font,
    colors="black",
    pad=7
)

ax.tick_params(
    axis="y",
    which="major",
    direction="in",
    left=True,
    right=True,
    length=4.9,
    width=1.1,
    labelsize=tick_font,
    colors="black",
    pad=7
)

ax.tick_params(
    axis="y",
    which="minor",
    direction="in",
    left=True,
    right=True,
    length=2.8,
    width=0.8,
    colors="black"
)

for spine in ax.spines.values():
    spine.set_color("black")
    spine.set_linewidth(1.1)


fig.subplots_adjust(
    left=0.18,
    right=0.97,
    bottom=0.18,
    top=0.96
)

fig.savefig("catalyst_amount_comparison.pdf",
    format="pdf"
)

plt.show()
```
