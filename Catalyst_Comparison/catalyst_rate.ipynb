```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import scienceplots

from pathlib import Path
from matplotlib.ticker import LogLocator, FuncFormatter
from matplotlib.patches import Patch


plt.style.use(["science", "ieee"])

tick_font       = 24.0
axis_label_font = 26.5
legend_font     = 22.0
value_font      = 22.0

mpl.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Times"],
    "font.size": tick_font,
    "axes.labelsize": axis_label_font,
    "xtick.labelsize": tick_font,
    "ytick.labelsize": tick_font,
    "legend.fontsize": legend_font,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "text.color": "black",
    "axes.labelcolor": "black",
    "axes.edgecolor": "black",
    "xtick.color": "black",
    "ytick.color": "black",
    "axes.grid": False,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "pdf.fonttype": 42,
    "ps.fonttype": 42
})


catalyst_labels = [
    "Ni\n600 $^\\circ$C",
    "Fe\n725 $^\\circ$C",
    "AC\n825 $^\\circ$C"
]

# Catalyst flow rates in kg/h
cstr_values = np.array([
    101.31,
    8610.60,
    64061.70
], dtype=float)

pfr_values = np.array([
    58.37,
    7915.14,
    62588.18
], dtype=float)


cstr_value_labels = [
    "101.3",
    "8,610.6",
    "70,490.8"
]

pfr_value_labels = [
    "58.4",
    "7,915.1",
    "61,838.3"
]

color_Ni = "#E69F00"
color_Fe = "#7B4AB5"
color_AC = "#D62728"

catalyst_colors = [
    color_Ni,
    color_Fe,
    color_AC
]

#Log scale
y_min = 20
y_max = 180_000


fig, ax = plt.subplots(
    figsize=(11.2, 7.1)
)

fig.patch.set_facecolor("white")
ax.set_facecolor("white")


x_positions = np.array([
    0.0,
    1.60,
    3.20
], dtype=float)

bar_width = 0.54


#CSTR
bars_CSTR = ax.bar(
    x_positions - bar_width / 2,
    cstr_values - y_min,
    bottom=y_min,
    width=bar_width,
    color=catalyst_colors,
    edgecolor="black",
    linewidth=1.1
)


#PFR
bars_PFR = ax.bar(
    x_positions + bar_width / 2,
    pfr_values - y_min,
    bottom=y_min,
    width=bar_width,
    color=catalyst_colors,
    edgecolor="black",
    linewidth=1.1,
    hatch="//"
)


ax.set_yscale("log")

ax.set_ylim(
    y_min,
    y_max
)


ax.yaxis.set_major_locator(
    LogLocator(
        base=10,
        numticks=10
    )
)

ax.yaxis.set_minor_locator(
    LogLocator(
        base=10,
        subs=np.arange(2, 10) * 0.1,
        numticks=100
    )
)


ax.yaxis.set_major_formatter(
    FuncFormatter(
        lambda value, position: f"{value:,.0f}"
    )
)

ax.yaxis.set_minor_formatter(
    FuncFormatter(
        lambda value, position: ""
    )
)



cstr_label_factors = [
    1.08,
    1.16,
    1.15
]

pfr_label_factors = [
    1.10,
    1.08,
    1.08
]


for bar, value, value_label, factor in zip(
    bars_CSTR,
    cstr_values,
    cstr_value_labels,
    cstr_label_factors
):

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value * factor,
        value_label,
        ha="center",
        va="bottom",
        fontsize=value_font,
        color="black"
    )


for bar, value, value_label, factor in zip(
    bars_PFR,
    pfr_values,
    pfr_value_labels,
    pfr_label_factors
):

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value * factor,
        value_label,
        ha="center",
        va="bottom",
        fontsize=value_font,
        color="black"
    )


ax.set_ylabel(
    r"Catalyst flow rate / "
    r"[kg$\cdot$h$^{\mbox{-}1}$]",
    fontsize=axis_label_font,
    labelpad=12
)


#X axis

ax.set_xticks(
    x_positions
)

ax.set_xticklabels(
    catalyst_labels,
    fontsize=tick_font
)


ax.tick_params(
    axis="both",
    which="major",
    direction="in",
    top=True,
    right=True,
    length=8,
    width=1.2,
    labelsize=tick_font,
    pad=8
)

ax.tick_params(
    axis="both",
    which="minor",
    direction="in",
    top=True,
    right=True,
    length=4,
    width=0.9
)

for spine in ax.spines.values():
    spine.set_linewidth(1.2)
    spine.set_color("black")


legend_handles = [
    Patch(
        facecolor="white",
        edgecolor="black",
        linewidth=1.0,
        label="CSTR"
    ),

    Patch(
        facecolor="white",
        edgecolor="black",
        linewidth=1.0,
        hatch="//",
        label="PFR"
    )
]

legend = ax.legend(
    handles=legend_handles,
    frameon=True,
    fancybox=False,
    framealpha=1.0,
    facecolor="white",
    edgecolor="0.50",
    fontsize=legend_font,
    loc="upper left",
    bbox_to_anchor=(0.03, 0.98),
    borderpad=0.35,
    handlelength=1.8,
    handletextpad=0.5,
    labelspacing=0.35
)

legend.get_frame().set_linewidth(0.8)

fig.subplots_adjust(
    left=0.13,
    right=0.985,
    bottom=0.20,
    top=0.96
)


fig.savefig("catalyst_flow_rate_cstr_vs_pfr.pdf",
    format="pdf"
)


plt.show()
```
